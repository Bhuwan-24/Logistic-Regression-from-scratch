```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import random
from sklearn.model_selection import train_test_split as tt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,recall_score,precision_score,confusion_matrix,f1_score
```

```python
class Logistic_Regression:

    def __init__(self,Learning_Rate=0.1,epochs=3000):
        self.weight=None
        self.intercept=None
        self.lr=Learning_Rate
        self.epochs=epochs

    def sigmoid(self,z):
        z=np.clip(z,-500,500)
        return 1/(1+np.exp(-z))

    def fit(self,X,Y):
        X=np.insert(X,0,1,axis=1)
        self.weight=np.ones(X.shape[1])

        for i in range(self.epochs):
            indx=random.randint(0,X.shape[0]-1)
            y_pred=self.sigmoid(np.dot(X[indx],self.weight))
            self.weight-=self.lr*(y_pred-Y[indx])*X[indx]
            self.intercept=self.weight[0]
            self.coef=self.weight[1:]

    def predict(self,x):
        x=np.insert(x,0,1,axis=1)
        self.z= np.dot(x,self.weight)
        self.y=self.sigmoid(self.z)
        return (self.y>0.5).astype(int)
```

```python
df=pd.read_csv("Social_Network_Ads.csv")
df['Gender']=df['Gender'].replace({'Male':1,'Female':0})
X=df[['Gender','Age','EstimatedSalary']].values
X=(X-np.mean(X))/np.std(X)
Y=df['Purchased'].values
x_train,x_test,y_train,y_test=tt(X,Y,test_size=0.2,random_state=2)
```

```python
lr=Logistic_Regression()
lr.fit(x_train,y_train)
lr.predict(x_test)
```

```python
model=LogisticRegression()
model.fit(x_train,y_train)
model.predict(x_test)
```

```python
print(model.intercept_,model.coef_)
print(lr.intercept,lr.coef)
```

```python
print(accuracy_score(lr.predict(x_test),y_test))
print(accuracy_score(model.predict(x_test),y_test))
```

```python
print(recall_score(lr.predict(x_test),y_test))
print(recall_score(model.predict(x_test),y_test))
```

```python
print(precision_score(lr.predict(x_test),y_test))
print(precision_score(model.predict(x_test),y_test))
```

```python
print(confusion_matrix(lr.predict(x_test),y_test))
print(confusion_matrix(model.predict(x_test),y_test))
```

```python
print(f1_score(lr.predict(x_test),y_test))
print(f1_score(model.predict(x_test),y_test))
```
