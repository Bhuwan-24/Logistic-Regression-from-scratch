```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split as tt
```

```python
class Logistic_Regression:
    def __init__(self,LearningRate=0.01,epochs=1000):
        self.lr=LearningRate
        self.w=None
        self.intercept=None
        self.coef=None
        self.epochs=epochs

    def sigmoid(self,z):
        return 1/(1+np.exp(-z))

    def fit(self,X,Y):
        X=np.insert(X,0,1,axis=1)
        self.w=np.ones(X.shape[1])
        m=X.shape[0]

        for i in range(self.epochs):
            y_pred=self.sigmoid(np.dot(X,self.w.T))
            self.w=self.w-(self.lr*np.dot(X.T,(Y-y_pred)))/m
            self.intercept=self.w[0]
            self.coef=self.w[1:]


    def predict(self,x):
        x=np.insert(x,0,1)
        ans=self.sigmoid(np.dot(x,self.w.T))
        return 1 if ans>0.5 else 0
```

```python
df=pd.read_csv('Social_Network_Ads.csv')
df['Gender']=df['Gender'].replace({'Male':1,'Female':0})
X=df[['Gender','Age','EstimatedSalary']].values
Y=df['Purchased']
x_train,x_test,y_train,y_test=tt(X,Y,test_size=0.2,random_state=2)
```

```python
lr=Logistic_Regression()
lr.fit(x_train,y_train)
lr.predict(x_test)
```
