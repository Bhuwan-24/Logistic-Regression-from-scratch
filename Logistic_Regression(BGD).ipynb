```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
class Logistic_Regression:
    def __init__(self,LearningRate=0.01,epochs=1000):
        self.lr=LearningRate
        self.w=None
        self.intercept=None
        self.coef=None
        self.epochs=epochs

    def sigmoid(self,z):
        return 1/(1+np.exp(-z))

    def fit(self,X,Y):
        X=np.insert(X,0,1,axis=1)
        self.w=np.ones(X.shape[1])
        prev_w=0
        itr=0
        m=X.shape[0]

        for i in range(self.epochs):
            itr+=1
            prev_w=self.w
            y_pred=self.sigmoid(np.dot(X,self.w.T))
            self.w=self.w+(self.lr*(Y-y_pred)*X)/m
            self.intercept=self.w[0]
            self.coef=self.w[1:]


    def predict(self,x):
        x=np.insert(0,1,axis=1)
        ans=self.sigmoid(np.dot(x,self.w.T))
        return 1 if ans>0.5 else 0
            

```
