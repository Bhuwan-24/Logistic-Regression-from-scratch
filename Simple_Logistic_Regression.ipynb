```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import random
```

```python
class Logistic_Regression:

    def __init__(self,Learning_Rate=0.01,epochs=100):
        self.weight=None
        self.intercept=None
        self.lr=Learning_Rate
        self.epochs=epochs

    def sigmoid(self,z):
        return 1/(1+np.exp(-z))

    def fit(self,X,Y):
        X=np.insert(X,0,1,axis=1)
        self.weight=np.ones(X.shape[1])

        for i in range(self.epochs):
            indx=random.randint(0,X.shape[0]-1)
            y_pred=self.sigmoid(np.dot(X[indx],self.weight))
            self.weight=self.weight-self.lr*(Y[indx]-y_pred)*X[indx]
            self.intercept=self.weight[0]
            self.coef=self.weight[1:]

    def predict(self,x):
        z= np.dot(self.coef.T,x)+self.intercept
        Y=self.sigmoid(z)
        return 1 if Y>0.5 else 0
```
