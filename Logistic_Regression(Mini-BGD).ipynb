```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
import random
```

```python
class Logistic_Regression():

    def __init__(self,LearningRate=0.1,epochs=1000,batch_size=10):
        self.lr=LearningRate
        self.epochs=epochs
        self.weight=None
        self.coef=None
        self.intercept=None
        self.size=batch_size

    def sigmoid(self,z):
        z=np.clip(z,-500,500)
        return 1/(1+np.exp(-z))

    def fit(self,X,Y):
        X=np.insert(X,0,1,axis=1)
        self.weight=np.ones(X.shape[1])

        for _ in range(self.epochs):
            indx=random.sample(range(X.shape[0]-1),self.size)
            y_pred=self.sigmoid(np.dot(X[indx],self.weight.T))
            err=(y_pred-Y[indx])
            grad=np.dot(X[indx].T,err)/self.size
            self.weight-=self.lr*grad
            self.intercept=self.weight[0]
            self.coef=self.weight[1:]

    def predict(self,x):
        x=np.insert(x,0,1,axis=1)
        pred=self.sigmoid(np.dot(x,self.weight))
        return (pred>0.5).astype(int)
```
