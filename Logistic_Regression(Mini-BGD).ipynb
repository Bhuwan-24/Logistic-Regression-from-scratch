```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,precision_score
from sklearn.model_selection import train_test_split as tt
import random
```

```python
class Logistic_Regression():

    def __init__(self,LearningRate=0.1,epochs=1000,batch_size=10):
        self.lr=LearningRate
        self.epochs=epochs
        self.weight=None
        self.coef=None
        self.intercept=None
        self.size=batch_size

    def sigmoid(self,z):
        z=np.clip(z,-500,500)
        return 1/(1+np.exp(-z))

    def fit(self,X,Y):
        X=np.insert(X,0,1,axis=1)
        self.weight=np.ones(X.shape[1])

        for _ in range(self.epochs):
            indx=random.sample(range(X.shape[0]-1),self.size)
            y_pred=self.sigmoid(np.dot(X[indx],self.weight.T))
            err=(y_pred-Y[indx])
            grad=np.dot(X[indx].T,err)/self.size
            self.weight-=self.lr*grad
            self.intercept=self.weight[0]
            self.coef=self.weight[1:]

    def predict(self,x):
        x=np.insert(x,0,1,axis=1)
        pred=self.sigmoid(np.dot(x,self.weight))
        return (pred>0.5).astype(int)
```

```python
df=pd.read_csv('Social_Network_Ads.csv')
df['Gender']=df['Gender'].replace({'Male':1,'Female':0})
X=df[['Gender','Age','EstimatedSalary']].values
Y=df['Purchased'].values
X=(X-np.mean(X))/np.std(X)
x_train,x_test,y_train,y_test=tt(X,Y,test_size=0.2,random_state=2)
```

```python
lr=Logistic_Regression()
lr.fit(x_train,y_train)
lr.predict(x_test)
```

```python
model=LogisticRegression()
model.fit(x_train,y_train)
model.predict(x_test)
```

```python
print(accuracy_score(model.predict(x_test),y_test))
print(accuracy_score(lr.predict(x_test),y_test))
```

```python
print(lr.intercept,lr.coef)
print(model.intercept_,model.coef_)
```

```python
print(precision_score(lr.predict(x_test),y_test))
print(precision_score(model.predict(x_test),y_test))
```
